```python
import sympy
```

```python
#symbolic derivatives
J, w = sympy.symbols('J,w')
```

```python

#when J=w^2

J = w**2
J
```

```python
dj_dw = sympy.diff(J,w)
dj_dw
```

```python
#plug-in a value of w=2
dj_dw.subs([(w,2)])
```

```python
dj_dw.subs([(w,3)])  #derivative at point w=3
```

```python
dj_dw.subs([(w, -3)]) #derivative at the point w=-3
```

```python
#when J=2w
```

```python
w, J = symbols('w, J')
```

```python
J = 2*w
J
```

```python
dj_dw = diff(J,w)
dj_dw
```

```python
dj_dw.subs([(w, -3)])  #derivative at the point w=-3
```

```python
#numpy arithmetic and sympy comparisons
J = 2*3
J_epsilon = 2*(3 + 0.001)
k = (J_epsilon - J)/ 0.001
print(f'J = {J}, J_epsilon = {J_epsilon}, dj_dw ~= k = {k}')
```

```python
#in the above we see any change in w will result in 2 times that amount of change in the output J, 
# regardless of the starting values
```

```python
#derivative describes how the output of a function changes when ther is 
# a small change in an inout value (epsilon~0.001)
#calculating J(w) = w^2 where w=3, e=0.001
J = (3)**2
J_epsilon = (3+0.001)**2
k = (J_epsilon - J) / 0.001  #difference divided by epsilon
print(f'J = {J}, J_epsilon = {J_epsilon}, dj_dw~= k = {k:0.6f}')
```

```python
#make epsilon really smaller = 0.000000001
J = (3)**2
J_epsilon = (3 + 0.000000001)**2
print(f'J = {J}, J_epsilon = {J_epsilon}, dj_dw ~= k = {k}')
```

```python
#the value gets closer to exactly 6 as we reduce the size of epsilon(change)
```
