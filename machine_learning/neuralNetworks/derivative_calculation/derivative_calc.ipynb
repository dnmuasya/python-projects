```python
import sympy
```

```python
J, w = sympy.symbols('J,w')
```

```python
#when J=w^2
J = w**2
J
```

```python
dj_dw = sympy.diff(J,w)
dj_dw
```

```python
#plug-in a value of w=2
dj_dw.subs([(w,2)])
```

```python
#if J was w^3
J = w**3
J
```

```python
dj_dw = sympy.diff(J, w)
dj_dw
```

```python
dj_dw.subs([(w,2)])
```

```python
#if J=w
J=w
J
```

```python
dj_dw = sympy.diff(J,w)
dj_dw
```

```python
dj_dw.subs([(w,2)])
```
