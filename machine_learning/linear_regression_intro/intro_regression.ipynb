```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('./deeplearning.mplstyle')
```

```python
''' 
An example of linear regression prediction model with one variable
The feature is the house size, and the target is House price
Use a simple data set with only two data points - a house with 1000sqft sold for $300000, 
and a house with 2000sqft sold for $500000. (Training set)
'''
```

```python
#x_train is the input variable(size in 1000 square feet)
#y_train is the output target(price in 1000s of dollars)
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
print(f'x_train = {x_train}')
print(f'y_train = {y_train}')
```

```python
#m is the number of training examples(entries)
print(f'x_train.shape: {x_train.shape}')
m = x_train.shape[0]
print(f'Number of training examples is: {m}')
```

```python
# alternatively use Python len() function to show the number of entries
m = len(x_train)
print(f'Number of training examples is: {m}')
```

```python
#You will use (x (𝑖) , y (𝑖) ) to denote the  𝑖𝑡ℎ   training example. Since Python is zero indexed, (x (0) , y (0)  )
#  is (1.0, 300.0) and (x (1) , y (1) ) is (2.0, 500.0).

#showing the training examples of the i-th value
#to access a value in a Numpy array, one indexes the array with the desired offset(change the offset to zero(0))
i = 1 # chang the offset to zero to see (x^0, y^0)
x_i = x_train[i]
y_i = y_train[i]
print(f'(x^({i}), y^({i})) = ({x_i}, {y_i}) ')
```

```python
#plotting the data
#you can user the `scatter()` function in the matplotlib library
# the function arguments `marker` and `c` show the points as red crosses(default is blue)
#plot the data points
plt.scatter(x_train, y_train, marker='x', c='r')

#set the title
plt.title('Housing Prices')

#set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')

#set the x-axis label
plt.xlabel('Size (1000 sqft)')

#show the plot
plt.show()
```

```python
w=200
b=100
print(f'w: {w}')
print(f'b: {b}')
```

```python
#let's compute the value of f(x^i) for the two data points.
#i.e x^0, f_wb = w*x[0] + b
# x^1, f_wb= w*x[1] + b

#using a for loop function
def compute_model_output(x, w, b):
    ''' 
    Computes the prediction of a linear model
    Args:
        x (ndarray (m,)): Data, m examples
        w, b (scalar)   : model parameters
    Returns
        f_wb(ndarray (m,)): model prediction

    '''
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range (m):
        f_wb[i] = w * x[i] + b

    return f_wb
```

```python
#let's call the `compute_model_output` function and plot out the output
from cProfile import label


tmp_f_wb = compute_model_output(x_train, w, b,)

#plot our model prediction
plt.plot(x_train, tmp_f_wb, c='b', label='Our Prediction')

#plot the data points
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual Values')

#set the title
plt.title('Housing Prices')

#set the y-label
plt.ylabel('Price (in 1000s of dollars)')

#set the x-label
plt.xlabel('Size (1000sqft)')

plt.legend()
plt.show()
```
